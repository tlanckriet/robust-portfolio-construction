### Why simple returns instead of log returns?

We use simple returns, $R_t = P_t / P_{t-1} - 1$, because portfolio optimisation combines returns
**across assets**. The return of a portfolio is the weighted sum of the simple returns of its components:

$$R_p = \sum_i w_i R_i$$

This does not hold for log returns, because the logarithm is not linear.

**Example:** €50 in asset A (+50%) and €50 in asset B (−50%). The portfolio ends at €100, a true return of 0%.
- Simple returns: $0.5 \times 50\% + 0.5 \times (-50\%) = 0\%$ ✓
- Log returns: $0.5 \times \ln(1.5) + 0.5 \times \ln(0.5) = -14.4\%$, which implies a portfolio return of −13.4% ✗

**Rule of thumb:**
- Simple returns aggregate **across assets** → used in portfolio construction ($w^\top\mu$, $w^\top\Sigma w$).
- Log returns aggregate **over time**: the log return over a year is the sum of the monthly log returns.

```python
rf = (monthly["^IRX"] / 100 / 12).shift(1)
rf = rf.loc[returns.index]
rf.head()
```

### Risk-free rate and look-ahead bias

We use the 13-week US T-bill yield (`^IRX`) as the risk-free rate. It is quoted as an
annualised percentage, so we convert it to a monthly decimal rate: $r_f = \text{IRX} / 100 / 12$.

We **lag the rate by one month** (`.shift(1)`): the risk-free return for month $t$ is the
yield observed at the end of month $t-1$.

- At the start of each month, an investor rebalances using only the information available
  at that moment: the most recent yield, from the end of the previous month.
- A T-bill bought at the start of the month locks in that yield, so it is also the
  risk-free return actually earned over the month.
- Using the end-of-month yield would introduce **look-ahead bias**: the backtest would rely
  on information that was not yet known when the investment decision was made.

```python
excess = returns.sub(rf, axis=0)

summary = pd.DataFrame({
    "Ann. return": returns.mean() * 12,
    "Ann. volatility": returns.std() * 12**0.5,
    "Sharpe": excess.mean() / excess.std() * 12**0.5,
})
summary.round(3)
```
